## To import Pandas library

```python
import pandas as pd
```

```python
df = pd.read_csv('data\cars.csv', engine='pyarrow')
```

## Data Cleaning

###  Find all Null Values in the dataset. If there is any null value in any column, then fill it with the mean of that column.

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df['Cylinders'].fillna(df['Cylinders'].mean(), inplace=True)
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

###  Check what are the different types of Make are there in our dataset. And, what is the count (occurrence) of each Make in the data?

```python
df['Make'].value_counts().head(10)
```

```python
df['Make'].value_counts().head(10).sort_values(ascending=True).plot(kind='barh')
```

###  Show all the records where Origin is Asia or Europe.

```python
df[df['Origin'].isin(['Asia', 'Europe'])].head(5)
```

### Remove all the records (rows) where Weight is above 4000.

```python
df[~(df['Weight'] > 4000)]
```

### Increase all the values of 'MPG_City' column by 3.

```python
df.head(2)
```

```python
df['MPG_City'] = df['MPG_City'].apply(lambda x: x + 3)
```

```python
df.head(2)
```

### Bar Graph

```python
df['Cylinders'] = df['Cylinders'].astype(int)
```

```python
df['Cylinders'].value_counts(ascending=True).plot(kind='barh')
```
